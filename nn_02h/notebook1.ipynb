```python
g = torch.Generator().manual_seed(42)
```

```python
# setup

xs, ys = [], []

for word in names:
    word = list("." + word + ".")
    for ch1, ch2 in zip(word, word[1:]):
        idx1 = stoi[ch1]
        idx2 = stoi[ch2]
        print(ch1, ch2)
        xs.append(idx1)
        ys.append(idx2)

xs = torch.tensor(xs)
ys = torch.tensor(ys)
xenc = F.one_hot(xs, num_classes=N).float()

W = torch.randn((N, N), generator=g, requires_grad=True)

# gradient descent training loop
for _ in range(100):

    # forward pass
    logits = xenc @ W
    counts = logits.exp()
    probs = counts / counts.sum(1, keepdim=True)

    # negative log likelihood loss
    loss = -probs[torch.arange(len(ys)), ys].log().mean()
    print(f"{loss=}")

    # backward pass
    W.grad = None
    loss.backward()

    # update weights (one step of gradient descent)
    learning_rate = 0.01
    W.data -= learning_rate * W.grad

print("Trained")
```

```python
# Sampling from the trained model

print("Samples")

for i in range(5):

    out = []
    idx = 0

    while True:

        xenc = F.one_hot(torch.tensor([idx]), num_classes=N).float()
        logits = xenc @ W
        counts = logits.exp()
        probs = counts / counts.sum(1, keepdim=True)

        idx = torch.multinomial(
            probs, num_samples=1, replacement=True, generator=g
        ).item()
        out.append(itos[idx])
        if idx == 0:
            break

    print("".join(out))
```
